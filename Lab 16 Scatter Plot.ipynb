
### Task 1: Exploring Relationships with Scatter Plots



#### 1. Scatter Plot for Age vs. Job Satisfaction



Visualize the relationship between respondents' age (`Age`) and job satisfaction (`JobSatPoints_6`). Use this plot to identify any patterns or trends.




```python
## Write your code here
# Map categorical age to numeric
age_mapping = {
    'Under 18 years old': 15,
    '18-24 years old': 21,
    '25-34 years old': 29.5,
    '35-44 years old': 39.5,
    '45-54 years old': 49.5,
    '55-64 years old': 59.5,
    '65 years or older': 70,
    'Prefer not to say': None
}
df['Age_Numeric'] = pd.to_numeric(df['Age'].map(age_mapping))

df_filtered = df[['Age_Numeric','JobSatPoints_6']].dropna()

# Remove top 1% outliers
q99 = df_filtered["Age_Numeric"].quantile(0.99)
df_filtered = df_filtered[df_filtered["Age_Numeric"] < q99]

plt.figure(figsize=(10, 6))
plt.scatter(df_filtered['Age_Numeric'],df_filtered['JobSatPoints_6'],
            alpha=0.7, c ='blue', marker = 'o')#, linewidths = 0.5, edgecolors = 'black')
plt.xlabel('Age')
plt.ylabel('JobSat')
plt.title('Scatter Plot for Age vs. Job Satisfaction')
plt.grid(True, linestyle = ':',linewidth = '1')
plt.tight_layout()
plt.show()
```

Add a regression line to the scatter plot of Age vs. JobSatPoints_6 to highlight trends in the data.


```python
## Write your code here
# Map categorical age to numeric
age_mapping = {
    'Under 18 years old': 15,
    '18-24 years old': 21,
    '25-34 years old': 29.5,
    '35-44 years old': 39.5,
    '45-54 years old': 49.5,
    '55-64 years old': 59.5,
    '65 years or older': 70,
    'Prefer not to say': None
}
df['Age_Numeric'] = pd.to_numeric(df['Age'].map(age_mapping))

df_filtered = df[['Age_Numeric','JobSatPoints_6']].dropna()

# Remove top 1% outliers
q99 = df_filtered["Age_Numeric"].quantile(0.99)
df_filtered = df_filtered[df_filtered["Age_Numeric"] < q99]

# Plot with regression line
plt.figure(figsize=(10,6))
sns.regplot(data=df_filtered, x='Age_Numeric', y='JobSatPoints_6', scatter_kws={'alpha':0.3}, line_kws={'color':'red','alpha':0.8})
plt.xlabel('Age')
plt.ylabel('Job Satisfaction (0–100)')
plt.title('Regression Line: Age vs. Job Satisfaction')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Task 3: Combining Scatter Plots with Additional Features


#### 1. Bubble Plot of Compensation vs. Job Satisfaction with Age as Bubble Size



Create a bubble plot to explore the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`), with bubble size representing age.


```python
help(sns.scatterplot)
```

```python
## Write your code here
# Map Age → Age_Numeric (already done earlier, uncomment if run alone)
# age_mapping = {
#     'Under 18 years old': 15,
#     '18-24 years old': 21,
#     '25-34 years old': 29.5,
#     '35-44 years old': 39.5,
#     '45-54 years old': 49.5,
#     '55-64 years old': 59.5,
#     '65 years or older': 70,
#     'Prefer not to say': None
# }
# df['Age_Numeric'] = pd.to_numeric(df['Age'].map(age_mapping))

# Drop missing
df_filtered = df[['Age_Numeric','JobSatPoints_6','ConvertedCompYearly']].dropna()
#Filter top 1% outliers
df_filtered = df_filtered[
    (df_filtered['Age_Numeric'] < df_filtered['Age_Numeric'].quantile(0.99)) &
    (df_filtered['ConvertedCompYearly'] < df_filtered['ConvertedCompYearly'].quantile(0.99))
]

# Create bubble plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_filtered,
    x='ConvertedCompYearly',
    y='JobSatPoints_6',
    size='Age_Numeric',         # Bubble size by Age
    sizes=(20, 300),            # Min and max bubble size
    hue='Age_Numeric',          # Color by Age
    palette='coolwarm',
    alpha=0.6,
    edgecolor='black',
    linewidth=0.3
)

plt.xlabel('Converted Compensation (Yearly)')
plt.ylabel('Job Satisfaction (0–100)')
plt.title('Bubble Plot: Compensation vs. Job Satisfaction (Bubble Size = Age)')
plt.grid(True, linestyle=':', linewidth=0.8)
plt.tight_layout()
plt.legend(title='Age (bubble size/color)', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

#### 2. Scatter Plot for Popular Programming Languages by Job Satisfaction


Visualize the popularity of programming languages (`LanguageHaveWorkedWith`) against job satisfaction using a scatter plot. Use points to represent satisfaction levels for each language.


```python
## Write your code here
# Split multiple languages per respondent
df_lang = df[['LanguageHaveWorkedWith', 'JobSatPoints_6']].dropna().copy()
df_lang['LanguageList'] = df_lang['LanguageHaveWorkedWith'].str.split(';')
df_exploded = df_lang.explode('LanguageList')
df_exploded['LanguageList'] = df_exploded['LanguageList'].str.strip()

# Optional: filter only popular languages
top_langs = df_exploded['LanguageList'].value_counts().nlargest(15).index
df_plot = df_exploded[df_exploded['LanguageList'].isin(top_langs)]

plt.figure(figsize=(14, 6))
sns.stripplot( #similar to a scatter plot but comes with a jitter param to partially show the overlap values
    data=df_plot,
    x='LanguageList',
    y='JobSatPoints_6',
    palette='Set3',
    legend = False, #use legend = Fasle and explicitly assign hue = x when use palette
    hue = 'LanguageList',
    jitter=True,
    alpha=0.7,
    edgecolor='auto', #auto is gray
    linewidth=0.4
)
plt.ylabel('Job Satisfaction (0–100)')
plt.xlabel('Programming Language')
plt.title('Job Satisfaction Across Programming Languages')
plt.grid(True, axis='y', linestyle=':', linewidth=0.8)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Task 4: Scatter Plot Comparisons Across Groups


#### 1. Scatter Plot for Compensation vs. Job Satisfaction by Employment Type


Visualize the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`), categorized by employment type (`Employment`). Use color coding or markers to differentiate between employment types.


```python
df['ConvertedCompYearly'].describe()
```

```python
df['JobSatPoints_6'].describe()
```

```python
df['Employment'].value_counts()
```

```python
## Write your code here
# Drop missing
df_filtered = df[['Employment','JobSatPoints_6','ConvertedCompYearly']].dropna()
#Filter top 1% outliers
df_filtered = df_filtered[df_filtered['ConvertedCompYearly'] < df_filtered['ConvertedCompYearly'].quantile(0.99)]

#Split employment to categories
df_filtered['Employment'] = df_filtered['Employment'].str.split(';')
df_filtered = df_filtered.explode('Employment').reset_index(drop = True)
df_filtered["Employment"] = df_filtered["Employment"].str.strip()

# Create plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_filtered,
    x='ConvertedCompYearly',
    y='JobSatPoints_6',
    hue='Employment',          # Color by Employment
    palette='coolwarm',
    #style='Employment',         #Marker style by Employment
    alpha=0.6,
    edgecolor='black',
    linewidth=0.3
)

plt.xlabel('Converted Compensation (Yearly)')
plt.ylabel('Job Satisfaction (0–100)')
plt.title('Compensation vs. Job Satisfaction by Empoyment Type')
plt.grid(True, linestyle=':', linewidth=0.8)
plt.tight_layout()
plt.legend(title='Employment', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

#### 2. Scatter Plot for Work Experience vs. Age Group by Country


Compare work experience (`YearsCodePro`) across different age groups (`Age`) and countries (`Country`). Use colors to represent different countries and markers for age groups.


```python
## Write your code here
#We dont need to convert Age to numeric as we're about to plot stringplot-look-a-like scatterplots, one for each country

# Step 1: Clean and convert YearsCodePro
def convert_years_code_pro(val):
    if val == 'Less than 1 year':
        return 0.5
    elif val == 'More than 50 years':
        return 51
    try:
        return float(val)
    except:
        return None

df['YearsCodePro_Num'] = df['YearsCodePro'].apply(convert_years_code_pro)

# Step 2: Drop missing values
df_clean = df[['YearsCodePro_Num', 'Age', 'Country']].dropna()

# Step 3: Get top 10 countries by count
top_countries = df_clean['Country'].value_counts().head(6).index
df_top = df_clean[df_clean['Country'].isin(top_countries)]

# Step 4: Plot with FacetGrid
g = sns.FacetGrid(df_top, col='Country', col_wrap=3, height=4, sharey=False)
g.map_dataframe(sns.scatterplot, x='Age', y='YearsCodePro_Num', palette='Set2',hue = 'Age', legend = False)
g.set_titles(col_template="{col_name}")
g.set_axis_labels("Age Group", "Years of Pro Coding")
for ax in g.axes.flatten():
    ax.tick_params(axis='x', rotation=45)
    ax.grid(True, linestyle='--', linewidth=1.2)
plt.tight_layout()
plt.subplots_adjust(top=0.92)
g.fig.suptitle("Years of Professional Coding Experience by Age Group (Top 5 Countries)")
plt.show()
```
