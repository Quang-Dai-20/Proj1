##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


Create a stacked chart to compare job satisfaction (`JobSatPoints_6` and `JobSatPoints_7`) across different employment statuses. This will show how satisfaction varies by employment type.


```python
##Write your code here
# 1. Preprocessing Employment
df_exploded = df.copy()
df_exploded['Employment'] = df_exploded['Employment'].str.split(';')
df_exploded = df_exploded.explode('Employment')

# 2. Group by Employment và tính tổng điểm hài lòng
agg_df = df_exploded.groupby('Employment')[['JobSatPoints_6', 'JobSatPoints_7']].mean()
agg_df['Total'] = agg_df['JobSatPoints_6'] + agg_df['JobSatPoints_7']
agg_df = agg_df.sort_values('Total', ascending=False)


# 3. Stacked bar chart
fig, ax = plt.subplots(figsize=(10, 6))

bars_6 = ax.bar(
    agg_df.index,
    agg_df['JobSatPoints_6'],
    label='Job Satisfaction (Q6)',
    color='skyblue'
)

bars_7 = ax.bar(
    agg_df.index,
    agg_df['JobSatPoints_7'],
    bottom=agg_df['JobSatPoints_6'],  # xếp chồng lên
    label='Job Satisfaction (Q7)',
    color='lightgreen'
)

# Annotate
for i, (index, row) in enumerate(agg_df.iterrows()):
    bottom = 0
    for col in agg_df.columns:
        value = row[col]
        ax.text(i, bottom + value / 2, f"{value:.2f}", ha='center', va='center', color='white', fontsize=10)
        bottom += value
        
ax.set_ylabel('Total Satisfaction Points')
ax.set_title('Job Satisfaction by Employment', fontsize=14, weight='bold')
ax.legend(loc='upper left', bbox_to_anchor=(1.0, 1.0))
plt.xticks(rotation=45, ha='right',)
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Preprocessing
df_clean = df.dropna(subset=['Age', 'DatabaseWantToWorkWith']).copy()
df_clean['DatabaseWantToWorkWith'] = df_clean['DatabaseWantToWorkWith'].str.split(';')
df_exploded = df_clean.explode('DatabaseWantToWorkWith')
df_exploded['DatabaseWantToWorkWith'] = df_exploded['DatabaseWantToWorkWith'].str.strip()

# 2. Select top N databases
top_n = 5
top_dbs = df_exploded['DatabaseWantToWorkWith'].value_counts().head(top_n).index
df_filtered = df_exploded[df_exploded['DatabaseWantToWorkWith'].isin(top_dbs)]

# 3. Group and pivot
grouped = df_filtered.groupby(['Age', 'DatabaseWantToWorkWith']).size().unstack(fill_value=0)

# 4. Plot stacked bar
ax = grouped.plot(kind='bar', stacked=True, figsize=(12, 8), colormap='Set3')

# 5. Annotate
for i, age in enumerate(grouped.index):
    y_offset = 0
    for db in grouped.columns:
        value = grouped.loc[age, db]
        if value > 0:
            ax.text(i, y_offset + value / 2, str(value), ha='center', va='center', fontsize=8)
            y_offset += value

# 6. Final touches
ax.set_ylabel('Number of Respondents')
ax.set_xlabel('Age Group')
ax.set_title('Top Databases People Want to Learn by Age Group')
plt.legend(title='Database', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
